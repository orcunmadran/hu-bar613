```python
import sqlite3

db_path = "/content/drive/MyDrive/Colab Notebooks/BAR613/chinook.db"
conn = sqlite3.connect(db_path)
```

```python
sql_query = f"SELECT * FROM artists"
cursor = conn.cursor()
cursor.execute(sql_query)
results = cursor.fetchall()
if results:
  for row in results:
    print(row)
else:
    print("Hiç kayıt yok..!")
```

```python
keyword = input("Aranacak anahtar kelimeyi girin: ")
sql_query = f"SELECT * FROM artists WHERE Name LIKE '%{keyword}%'"
cursor = conn.cursor()
cursor.execute(sql_query)
results = cursor.fetchall()
if results:
    print(f"'{keyword}' içeren sanatçılar:")
    for row in results:
        print(row)
else:
    print(f"'{keyword}' içeren sanatçı bulunamadı.")
```

```python
cursor.close()
conn.close()
```
